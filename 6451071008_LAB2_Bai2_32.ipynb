1. Tính F-measure từ Precision và Recall

```python
print("Họ tên: Trần Thị Mỹ Dung")
print("MSSV: 6451071008")

def tinh_f_measure(precision, recall):
    if precision + recall == 0:
        return 0
    return 2 * precision * recall / (precision + recall)

ket_qua = {
    "ID3": (0.85, 0.80),
    "CART": (0.82, 0.78),
    "Naive Bayes": (0.80, 0.75)
}

for ten, (precision, recall) in ket_qua.items():
    f1 = tinh_f_measure(precision, recall)

    print("\nMô hình:", ten)
    print("Precision:", precision)
    print("Recall:", recall)
    print("F-measure:", round(f1, 4))
```

2. Tính F-measure trực tiếp từ ma trận nhầm lẫn

```python
import numpy as np
from sklearn.metrics import precision_score, recall_score, f1_score

print("Họ tên: Trần Thị Mỹ Dung")
print("MSSV: 6451071008")

# Ma trận nhầm lẫn
cm = np.array([
    [2, 0],
    [0, 1]
])

print("Ma trận nhầm lẫn:")
print(cm)

# Lấy TP, FP, FN từ ma trận nhầm lẫn
TN, FP, FN, TP = cm.ravel()

# Tính Precision
precision = TP / (TP + FP) if (TP + FP) != 0 else 0

# Tính Recall
recall = TP / (TP + FN) if (TP + FN) != 0 else 0

# Tính F-measure
f_measure = (
    2 * precision * recall / (precision + recall)
    if (precision + recall) != 0 else 0
)

print("\nPrecision:", precision)
print("Recall:", recall)
print("F-measure:", f_measure)
```
